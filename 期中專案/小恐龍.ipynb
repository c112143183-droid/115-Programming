```python
import time
import cv2
import numpy as np
import pyautogui as pyg
import mss

pyg.PAUSE = 0.0
pyg.FAILSAFE = True

# ==============================
# 1. 螢幕偵測範圍設定
# ==============================

game_area = {
    "left": 285,
    "top": 390,
    "width": 335,
    "height": 60
}

base_x = 240
base_width = 29

obstacle_area = {
    "left": base_x,
    "top": 485,
    "width": base_width,
    "height": 45
}

# ==============================
# 2. 狀態與速度補償設定
# ==============================

last_obstacle = False
clear_count = 0
jump_count = 0
start_time = None

SPEED_FACTOR = 1.9
MAX_EXTEND_X = 130

CLEAR_FRAMES = 4
JUMP_HOLD = 0.11  # 原本 0.08，本次只調整這裡

print("超低延遲版恐龍跳躍啟動！")
print("3 秒後開始偵測，請切換至 Chrome 恐龍視窗...")
time.sleep(3)

start_time = time.time()

# ==============================
# 3. 高速主循環
# ==============================

try:
    with mss.MSS() as sct:
        frame_idx = 0

        while True:
            frame_idx += 1

            # 遊戲結束判定
            if frame_idx % 10 == 0:
                game_img = np.array(sct.grab(game_area))

                if np.count_nonzero(
                    np.all(game_img[:, :, :3] < 150, axis=2)
                ) > 100:
                    print("\n遊戲結束！")
                    print(f"總計起跳次數：{jump_count}")
                    break

            # 動態調整偵測位置
            elapsed = time.time() - start_time
            offset = int(min(
                elapsed * SPEED_FACTOR,
                MAX_EXTEND_X
            ))

            cur_area = {
                "left": base_x + offset,
                "top": obstacle_area["top"],
                "width": obstacle_area["width"],
                "height": obstacle_area["height"]
            }

            # 擷取窄光閘畫面
            obstacle_img = np.array(sct.grab(cur_area))

            gray = cv2.cvtColor(
                obstacle_img,
                cv2.COLOR_BGRA2GRAY
            )

            _, binary = cv2.threshold(
                gray,
                180,
                255,
                cv2.THRESH_BINARY_INV
            )

            contours, _ = cv2.findContours(
                binary,
                cv2.RETR_EXTERNAL,
                cv2.CHAIN_APPROX_SIMPLE
            )

            obstacle_detected = False

            for contour in contours:
                x, y, w, h = cv2.boundingRect(contour)

                if 2 <= w <= 25 and 20 <= h <= 45:
                    obstacle_detected = True
                    break

            # 偵測到新障礙物就起跳
            if obstacle_detected and not last_obstacle:

                pyg.keyDown("space")
                time.sleep(JUMP_HOLD)
                pyg.keyUp("space")

                jump_count += 1
                last_obstacle = True
                clear_count = 0

                print(
                    f"第 {jump_count} 次起跳 | "
                    f"偵測位置：{cur_area['left']}"
                )

            # 障礙物離開後重新準備
            elif not obstacle_detected and last_obstacle:

                clear_count += 1

                if clear_count >= CLEAR_FRAMES:
                    last_obstacle = False
                    clear_count = 0

            time.sleep(0.005)

finally:
    pyg.keyUp("space")
    print("程式已停止。")
```
